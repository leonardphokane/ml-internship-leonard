## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

## Rule (plain words):  
Pick actions that show consistent positive impact on the baseline metric. Score them by how strong and reliable the effect is.

## Reason codes (outputs):

**HIT** — clear positive effect, repeatable

**MISS** — no measurable effect

**NOISE** — inconsistent or too small to trust

**LEAK** — uses future info or product flag (invalid)

**WEAK** — looks positive but fragile

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

We compute a baseline action score for each action, then rank them. The output goes to work/outputs/baseline_action_score.csv.

```python
import os
import pandas as pd

# Create the outputs folder if it doesn't exist
output_dir = r"C:\Users\mokga\Projects\ml-internship-leonard\work\outputs"
os.makedirs(output_dir, exist_ok=True)

# Example data
data = [
    {"action":"A1","lift":0.12,"confidence":0.9,"reason":"HIT"},
    {"action":"A2","lift":0.05,"confidence":0.6,"reason":"WEAK"},
    {"action":"A3","lift":0.00,"confidence":0.5,"reason":"MISS"},
    {"action":"A4","lift":0.20,"confidence":0.95,"reason":"HIT"},
]

df = pd.DataFrame(data)
df["score"] = df["lift"] * df["confidence"]
df = df.sort_values("score", ascending=False).reset_index(drop=True)

# Save to the correct folder
csv_path = os.path.join(output_dir, "baseline_action_score.csv")
df.to_csv(csv_path, index=False)

print("File saved to:", csv_path)
print("Exists?", os.path.exists(csv_path))
```
