## Example Finding A: CTR vs Position

**Label source**: CTR is measured directly from click logs; position comes from ranking outputs.

**Validation design**: The paper used bucketed averages across positions. This design supports the claim because CTR consistently declines as position worsens.

Methodology question**: Did the validation control for query intent? CTR differences might partly reflect navigational vs informational queries.

## Example Finding B: Staleness and Refresh Flags

**Label source**: Staleness is derived from document age relative to last update; refresh flags are hand‑written rules.

**Validation design**: The paper compared stale vs fresh documents on engagement. The design carries the claim directionally, but it may conflate staleness with seasonal content.

**Methodology question**: Was staleness measured uniformly across domains (news vs evergreen)? Different domains age differently.
