## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

- **Unbalanced history**: Some clients or content IDs have far more rows than others, which skews observed patterns.  
- **Sparse early signals**: Initial rows may be Google Search Console (GSC)–only, missing richer engagement features.  
- **Window overlaps**: Refresh events can overlap across campaigns, making attribution noisy.  
- **Directional only**: Trends (`trend_pct`, `trend_direction`) are observed, not causal — they support decisions but don’t prove outcomes.  
- **Privacy constraints**: Client identifiers are anonymized, so the data cannot tell us about specific organizations or URLs.
