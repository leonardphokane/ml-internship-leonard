## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human Review Required:

Any recommendation with unusually high or low scores.

Actions that imply sensitive or client-specific content.

## No-Go (Never Automated):

Direct publishing of client-facing content.

Decisions involving legal, ethical, or compliance-sensitive material.

Any action where cost > measurable value.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Triggers for Monitoring:

Engagement drops >15% compared to baseline.

Archetype distribution shifts significantly (e.g., new dominant class).

Model drift detected in validation metrics.

## Retrain Triggers:

Data refresh cycle exceeds 8 weeks.

Accuracy falls below 70% on validation.

Human reviewers flag >10% of outputs as unreliable.
